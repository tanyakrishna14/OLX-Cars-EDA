```python
#importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#replacing all the empty values with the mode of that column
df["Car Profile"]=df["Car Profile"].fillna(df["Car Profile"].mode())
```

```python
#remove column named 'Description'
df=df.drop(columns=["Description"])
```

```python
#print all column names
df.columns
```

```python
#print numerical columns
df.select_dtypes(include=np.number)
```

```python
#print categorical columns
df.select_dtypes(include="object")
```

```python
#print unique values of every categorical column
for i in df.select_dtypes(include="object"):
    print(i)
    print(df[i].unique())
    print()
```

```python
#print frequency of every category
for i in df.select_dtypes(include="object"):
    print(df[i].value_counts())
    print()
```

```python
#boxplot for numerical columns
for i in df.select_dtypes(include=np.number):
    sns.boxplot(x=df[i])
    plt.title(i)
    plt.show()
```

```python
#remove outliers
Q1=df["Price"].quantile(0.25)
Q3=df["Price"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
df=df[(df["Price"]>=lower) & (df["Price"]<=upper)]
```

```python
#univariate
sns.histplot(df["Fuel"])
plt.show()
```

```python
#Transmission
sns.countplot(x="Transmission",data=df)
plt.title("Transmission Distribution")
plt.show()
```

```python
#Car documents
sns.countplot(y="Car documents",data=df)
plt.title("Car documents")
plt.show()
```

```python
#Price
sns.histplot(df["Price"],bins=20)
plt.title("Price Distribution")
plt.show()
```

```python
#Year
sns.histplot(df["Year"])
plt.title("Manufacturing Year")
plt.show()
```

```python
#bivariate
#Fuel vs Transmission
sns.countplot(x="Fuel",hue="Transmission",data=df)
plt.show()
```

```python
#Fuel vs Price
sns.boxplot(x="Fuel",y="Price",data=df)
plt.title("Fuel vs Price")
plt.show()
```

```python
#Year vs Price
sns.scatterplot(x="Year",y="Price",data=df)
plt.title("Year vs Price")
plt.show()
```

```python
#multivariate
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
